# Module 2 Assignment: Salary Function

```python
# Import the required modules
import pandas as pd
import csv
import shutil
from pathlib import Path

# Load the data using pandas
try:
    data = pd.read_csv("Total.csv", dtype=str, na_values=["Not Provided"])
    if data.empty or "EmployeeName" not in data.columns:
        raise ValueError("The CSV must contain employees records including an expected EmployeeName column.")
except (OSError, ValueError, pd.errors.ParserError) as error:
    raise ValueError(f"An error occurred while loading salary data: {error}") from error

rows = data.fillna("").to_dict("records")
print(f"{len(rows)} records loaded.")
```

```python
# Create a case insensitive employee function which accepts an employee name and returns its records
def find_employee(name):
    """Return the salary records matching a given employee's name."""
    if not isinstance(name, str) or not name.strip():
        raise ValueError("Please enter an employee name.")
    name = " ".join(name.split()).lower()
    if name not in employees:
        raise ValueError("Employee not found.")
    return employees[name]
```

```python
# Data processing using dictionary approach
employees = {}
for row in rows:
    name = " ".join(row["EmployeeName"].split()).lower()
    if name:
        employees.setdefault(name, []).append(row)

print(f"Stored {len(employees):,} employee names.")

employee_name = "NATHANIEL FORD"  # You can Change this to look up another employee.
try:
    details = find_employee(employee_name)
    display(pd.DataFrame(details))
except ValueError as error:
    print(error)
```

```python
# Error handling
# Checks for empty, invalid and unknown names
for name in ["", None, "__UNKNOWN_EMPLOYEE__"]:
    try:
        find_employee(name)
    except ValueError as error:
        print(f"{name!r}: {error}")
```

```python
# Export the employee details / profile
try:
    details = find_employee(employee_name)
    folder = Path("Employee Profile")
    folder.mkdir(exist_ok=True)
    with open(folder / "employee_details.csv", "w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=data.columns)
        writer.writeheader()
        writer.writerows(details)

    shutil.make_archive("Employee Profile", "zip", root_dir=".", base_dir=str(folder))
    print("Saved Employee Profile.zip")
except (ValueError, OSError, csv.Error) as error:
    print(f"Could not save profile: {error}")
```
